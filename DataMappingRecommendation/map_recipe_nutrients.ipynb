```python
recipe_interaction_df = raw_recipes_df.merge(raw_interactions_df, left_on="id", right_on="recipe_id", how="left")
recipe_interaction_df
```

```python
recipe_names = {}
for recipe_id, recipe_name in raw_recipes_df[["id", "name"]].itertuples(index=False):
    recipe_names[recipe_id]=recipe_name
recipe_names
```

```python
# user_id -> list of (recipe_id, rating, date)
# Keep the original interaction date with each event; repeated user/recipe
# ratings can have different dates and must remain separate rows.
user_ratings = {}
for user_id, recipe_id, rating, date in raw_interactions_df[
    ["user_id", "recipe_id", "rating", "date"]
].itertuples(index=False):
    user_ratings.setdefault(user_id, []).append((recipe_id, rating, date))

user_ratings
```

```python
# Reference 100% Daily Value (DV) values based on standard FDA 2,000-calorie daily diet guidelines
DAILY_VALUES_100 = [
    2000.0,  # [0] Calories (kcal)
    65.0,    # [1] Total Fat (grams)
    50.0,    # [2] Sugar (grams)
    2400.0,  # [3] Sodium (milligrams)
    50.0,    # [4] Protein (grams)
    20.0,    # [5] Saturated Fat (grams)
    300.0    # [6] Carbohydrates (grams)
]
```

```python
import ast

nutrition_cols = [
    "calories (g)",
    "total fat (g)",
    "sugar (g)",
    "sodium (g)",
    "protein (g)",
    "saturated fat (g)",
    "carbohydrates (g)",
]

# recipe_id -> {nutrient_name: value}
recipe_nutrition = {}
# recipe_id -> [ingredients]
recipe_ingredients = {}

for recipe_id, nutrition, ingredients in raw_recipes_df[["id", "nutrition", "ingredients"]].itertuples(index=False):
    nutrition_values = ast.literal_eval(nutrition)
    for i in range(len(nutrition_values)):
        nutrition_values[i]=nutrition_values[i]*DAILY_VALUES_100[i]       # the number in dataset is percentage, so we have to to get the actual value of nutrition value

    ingredients_list = ast.literal_eval(ingredients)
    recipe_nutrition[recipe_id] = dict(zip(nutrition_cols, nutrition_values))
    recipe_ingredients[recipe_id] = ingredients_list

recipe_nutrition, recipe_ingredients
```

```python
import os

# combine the 3 dicts into one recipe-centric dataframe
recipe_df = pd.DataFrame(
    [
        {
            "recipe_id": recipe_id,
            "name": recipe_names[recipe_id],
            **recipe_nutrition.get(recipe_id, {}),
            "ingredients": recipe_ingredients.get(recipe_id, []),
        }
        for recipe_id in recipe_names
    ]
)

# Flatten complete interaction events, including their original CSV date.
user_rating_df = pd.DataFrame(
    [(user_id, recipe_id, rating, date)
     for user_id, vals in user_ratings.items()
     for recipe_id, rating, date in vals],
    columns=["user_id", "recipe_id", "rating", "date"],
)

# Keep every recipe and every interaction. Recipes without interactions keep
# a missing date, rather than borrowing the recipe's submission date.
final_df = recipe_df.merge(
    user_rating_df, on="recipe_id", how="left", validate="one_to_many", sort=False
)

os.makedirs("input_stage2", exist_ok=True)
# recipe_user_ratings.csv is written by the final export cell, after the
# ingredient NER section, so the file also carries product/adj/verb columns.

final_df
```

```python
# Full singularization rules copied from 3_process_price.ipynb.
import re

# --- plural -> singular for the extracted product spans --------------------
# The NER hands back whatever the source text wrote, so "Apple" and "Apples"
# arrive as two different items. Every word of a span is put in the singular.

# The outliers first: the plurals the rules below get wrong.
IRREGULAR_PLURALS = {
    # classic irregulars
    "geese": "goose", "mice": "mouse", "lice": "louse", "teeth": "tooth",
    "feet": "foot", "children": "child", "men": "man", "women": "woman",
    "people": "person", "oxen": "ox", "testes": "testis",
    # -f / -fe stems: no rule separates "loaves" from "olives"
    "loaves": "loaf", "leaves": "leaf", "halves": "half", "knives": "knife",
    "calves": "calf", "hooves": "hoof", "shelves": "shelf", "sheaves": "sheaf",
    "lives": "life", "wolves": "wolf", "thieves": "thief", "scarves": "scarf",
    # the singular already ends in -ie / -i, so -ies -> -y would mangle it
    "pies": "pie", "cookies": "cookie", "brownies": "brownie",
    "smoothies": "smoothie", "veggies": "veggie", "hoagies": "hoagie",
    "toasties": "toastie", "smokies": "smokie", "crunchies": "crunchie",
    "calories": "calorie", "pierogies": "pierogi",
    "chilies": "chili", "chillies": "chilli",
    # the singular already ends in -oe / -e, so -oes -> -o would mangle it
    "toes": "toe", "roes": "roe", "aloes": "aloe", "shoes": "shoe",
    "canoes": "canoe", "mousses": "mousse",
    # a bare -s stem, so the -es rule would leave a trailing -e
    "buses": "bus", "gases": "gas", "octopuses": "octopus",
    # latinate and spanish plurals
    "cacti": "cactus", "fungi": "fungus", "octopi": "octopus",
    "larvae": "larva", "pasteles": "pastel", "nopales": "nopal",
}

# Ends in -s, but is already singular or has no singular at all.
UNCOUNTABLE = {
    # -s words that are not plurals
    "molasses", "series", "species", "news", "haggis", "pancreas", "gras",
    "always", "tennis", "cordyceps", "militaris",
    # mass nouns kept in the plural
    "oats", "grits", "greens", "goods", "brussels", "gyros", "tapas",
    # brands and place names that carry an -s of their own
    "mars", "texas", "snickers", "skittles", "twizzlers", "kudos", "krispies",
}

# The regular endings, in order; the first rule that matches wins.
PLURAL_RULES = [
    (re.compile(r"([^aeiou])ies$", re.I), r"\1y"),     # berries -> berry
    (re.compile(r"(ss|sh|ch|x|z)es$", re.I), r"\1"),   # peaches -> peach, boxes -> box
    (re.compile(r"([^aeiou])oes$", re.I), r"\1o"),     # tomatoes -> tomato
    (re.compile(r"([^s])s$", re.I), r"\1"),            # apples -> apple, cheeses -> cheese
]

WORD_RE = re.compile(r"[A-Za-z]+")


def _restore_case(word, singular):
    """Only the ending changed, so the stem keeps the spelling it came with."""
    low = word.lower()
    k = 0
    while k < len(singular) and low[k] == singular[k]:
        k += 1
    tail = singular[k:]
    if word.isupper() or word[k : k + 1].isupper():
        tail = tail.upper()
    return word[:k] + tail


def singularize_word(word):
    low = word.lower()
    if low in IRREGULAR_PLURALS:
        return _restore_case(word, IRREGULAR_PLURALS[low])
    # no food plural is shorter than four letters, but "was" / "his" are not either
    if low in UNCOUNTABLE or len(low) < 4 or not low.endswith("s"):
        return word
    # "hummus", "citrus", "asparagus", "plus" — here -us is a singular ending
    if low.endswith("us"):
        return word
    for rule, repl in PLURAL_RULES:
        if rule.search(low):
            return _restore_case(word, rule.sub(repl, low))
    return word


def singularize(span):
    """Every word of the span in the singular; punctuation is left as it is."""
    return WORD_RE.sub(lambda m: singularize_word(m.group()), str(span))
# Snapshot of the same NLTK English stopwords (retaining no/nor/not).
# Embedded so reruns do not require a corpus download.
STOP_WORDS = set(['about', 'above', 'after', 'again', 'against', 'ain', 'all', 'am', 'an', 'and', 'any', 'are', 'aren', "aren't", 'as', 'at', 'be', 'because', 'been', 'before', 'being', 'below', 'between', 'both', 'but', 'by', 'can', 'couldn', "couldn't", 'did', 'didn', "didn't", 'do', 'does', 'doesn', "doesn't", 'doing', 'don', "don't", 'down', 'during', 'each', 'few', 'for', 'from', 'further', 'had', 'hadn', "hadn't", 'has', 'hasn', "hasn't", 'have', 'haven', "haven't", 'having', 'he', "he'd", "he'll", "he's", 'her', 'here', 'hers', 'herself', 'him', 'himself', 'his', 'how', "i'd", "i'll", "i'm", "i've", 'if', 'in', 'into', 'is', 'isn', "isn't", 'it', "it'd", "it'll", "it's", 'its', 'itself', 'just', 'll', 'ma', 'me', 'mightn', "mightn't", 'more', 'most', 'mustn', "mustn't", 'my', 'myself', 'needn', "needn't", 'now', 'of', 'off', 'on', 'once', 'only', 'or', 'other', 'our', 'ours', 'ourselves', 'out', 'over', 'own', 're', 'same', 'shan', "shan't", 'she', "she'd", "she'll", "she's", 'should', "should've", 'shouldn', "shouldn't", 'so', 'some', 'such', 'than', 'that', "that'll", 'the', 'their', 'theirs', 'them', 'themselves', 'then', 'there', 'these', 'they', "they'd", "they'll", "they're", "they've", 'this', 'those', 'through', 'to', 'too', 'under', 'until', 'up', 've', 'very', 'was', 'wasn', "wasn't", 'we', "we'd", "we'll", "we're", "we've", 'were', 'weren', "weren't", 'what', 'when', 'where', 'which', 'while', 'who', 'whom', 'why', 'will', 'with', 'won', "won't", 'wouldn', "wouldn't", 'you', "you'd", "you'll", "you're", "you've", 'your', 'yours', 'yourself', 'yourselves'])

def normalize_description(text):
    """The description as written, only cleaned: no tokenizer, no stemming."""
    s = str(text).lower()
    s = re.sub(r"\(.*?\)", " ", s)        # "(China)" is origin, not part of the name
    s = re.sub(r"[^a-z0-9%\s]", " ", s)   # commas included: they are only separators
    kept = [w for w in s.split() if w not in STOP_WORDS and w.strip("%")]
    return " ".join(kept)


def normalize_price_text(value):
    """Apply the reference cleaning and singularization to BOTH sides."""
    return singularize(normalize_description(value))

# Only preparation words are dropped; e.g. almond, peanut, unsalted and
# gluten-free remain required. Extend aliases explicitly after reviewing gaps.
PRICE_PREPARATION_WORDS = {"fresh", "chopped", "diced", "sliced", "peeled", "grated"}
INGREDIENT_PRICE_ALIASES = {
    "garlic clove": "garlic",
    "all purpose flour": "plain flour",
    "scallion": "spring onion",
    "cilantro": "coriander",
}

INGREDIENT_PRICE_ALIASES = {
    normalize_price_text(k): normalize_price_text(v)
    for k, v in INGREDIENT_PRICE_ALIASES.items()
}

def normalize_price_ingredient(ingredient):
    key = " ".join(
        word for word in normalize_price_text(ingredient).split()
        if word not in PRICE_PREPARATION_WORDS
    )
    return INGREDIENT_PRICE_ALIASES.get(key, key)

PRICE_UNIT_FACTORS = {
    "kg": (1000.0, "g"), "g": (1.0, "g"),
    "l": (1000.0, "ml"), "ml": (1.0, "ml"),
    "lb": (453.592, "g"), "oz": (28.3495, "g"),
}

def parse_price_quantity(quantity):
    """Return (pack magnitude, g/ml); reject count-only or ambiguous packs."""
    text = re.sub(r"\s+", "", str(quantity).lower()).replace("×", "x")
    text = re.sub(r"\((?:ctn|\d+(?:perpack|pcs?|s))\)", "", text)
    total, dimension = 0.0, None
    for part in text.split("+"):
        match = re.fullmatch(r"(?:(\d+)x)?(\d+(?:\.\d+)?)(kg|g|ml|l|lb|oz)", part)
        if not match:
            return None, None
        count, amount, unit = match.groups()
        factor, current_dimension = PRICE_UNIT_FACTORS[unit]
        if dimension is not None and dimension != current_dimension:
            return None, None
        magnitude = int(count or 1) * float(amount) * factor
        if magnitude <= 0:
            return None, None
        total += magnitude
        dimension = current_dimension
    return (total, dimension) if total > 0 else (None, None)
```

**BERT embeddings and DBSCAN candidate segregation**

The early FairPrice diagnostic still embeds every normalized ingredient phrase
with `google-bert/bert-base-uncased` and uses DBSCAN to constrain its fallback.

For the final recipe-to-`price_mapped_nutrients.csv` mapping, both entity
DataFrames are handled field by field:

1. Parse every vocabulary value in `product`, `adj`, and `verb` on both the
   recipe ingredient DataFrame and nutrient-price DataFrame.
2. Normalize every value, apply POS-aware WordNet lemmatization (`n`, `a`, or
   `v`), then run the existing singularizer. This makes forms such as
   `waffles` → `waffle`, `frozen` → `freeze`, and `ground` → `grind` before
   encoding.
3. Embed each distinct normalized vocabulary phrase once with the pinned
   `google-bert/bert-base-uncased` checkpoint. For each field, subtract the
   shared vocabulary mean before L2 normalization; this removes BERT's common
   vector direction so a cosine above `0.8` represents genuinely close terms.
   A row's field embedding is the L2-normalized mean of those corrected vectors.
4. Run DBSCAN on the combined unique **product-field** vectors from both
   DataFrames. An unmatched recipe ingredient only compares with positively
   priced nutrient rows in its product cluster.
5. Require product cosine similarity above `0.8`. Rank candidates with a
   weighted score: product `0.70`, adjective `0.15`, verb `0.15`; an optional
   field participates only when both rows contain it. The combined score must
   also be above `0.8`.

Exact matches remain exact and fuzzy matches never become anchors for another
match. DBSCAN label `-1` means noise, so those rows remain unmatched after a
failed exact lookup. Embedding caches stay outside the two recipe CSVs.

```python
from pathlib import Path
from hashlib import sha256
import json
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
from transformers import AutoModel, AutoTokenizer
from huggingface_hub import try_to_load_from_cache

NUTRIENT_BERT_MODEL = "google-bert/bert-base-uncased"
NUTRIENT_BERT_REVISION = "86b5e0934494bd15c9632b12f734a8a67f723594"
NUTRIENT_BERT_THRESHOLD = 0.8  # acceptance uses >, never >=
NUTRIENT_BERT_BATCH_SIZE = 64
NUTRIENT_BERT_MAX_LENGTH = 128
nutrient_match_output_dir = Path("input_stage2")
nutrient_match_output_dir.mkdir(parents=True, exist_ok=True)
import tempfile
nutrient_cache_dir = (Path(tempfile.gettempdir()) / "recipe_price_mapping_cache"
                      / sha256(str(Path.cwd().resolve()).encode()).hexdigest()[:16])
nutrient_cache_dir.mkdir(parents=True, exist_ok=True)

def load_nutrient_bert():
    # Prefer the pinned requested checkpoint from the local cache.
    try:
        tokenizer = AutoTokenizer.from_pretrained(
            NUTRIENT_BERT_MODEL, revision=NUTRIENT_BERT_REVISION, local_files_only=True
        )
        model = AutoModel.from_pretrained(
            NUTRIENT_BERT_MODEL, revision=NUTRIENT_BERT_REVISION, local_files_only=True
        )
    except OSError:
        # The legacy repo ID refers to the identical BERT checkpoint. Reuse
        # its local weights only after verifying the official SHA256:
        # https://huggingface.co/google-bert/bert-base-uncased/blob/main/model.safetensors
        legacy_weight = try_to_load_from_cache(
            "bert-base-uncased", "model.safetensors", revision=NUTRIENT_BERT_REVISION
        )
        official_sha256 = "68d45e234eb4a928074dfd868cead0219ab85354cc53d20e772753c6bb9169d3"
        digest = None
        if isinstance(legacy_weight, str):
            with open(legacy_weight, "rb") as handle:
                weight_hash = sha256()
                for block in iter(lambda: handle.read(1024 * 1024), b""):
                    weight_hash.update(block)
            digest = weight_hash.hexdigest()
        if digest == official_sha256:
            tokenizer = AutoTokenizer.from_pretrained(
                "bert-base-uncased", revision=NUTRIENT_BERT_REVISION, local_files_only=True
            )
            model = AutoModel.from_pretrained(
                "bert-base-uncased", revision=NUTRIENT_BERT_REVISION, local_files_only=True
            )
            print("Reused local BERT weights; SHA256 matches google-bert/bert-base-uncased.")
        else:
            # On a new machine, cache the requested model in the temporary model cache.
            cache_dir = str(nutrient_cache_dir / "model_cache")
            tokenizer = AutoTokenizer.from_pretrained(
                NUTRIENT_BERT_MODEL, revision=NUTRIENT_BERT_REVISION, cache_dir=cache_dir
            )
            model = AutoModel.from_pretrained(
                NUTRIENT_BERT_MODEL, revision=NUTRIENT_BERT_REVISION, cache_dir=cache_dir
            )
    return tokenizer, model

def get_nutrient_bert_embeddings(texts):
    """Return sorted texts, unit BERT vectors, and their row lookup."""
    nutrient_embedding_texts = sorted(set(texts), key=lambda text: (len(text.split()), text))
    if not nutrient_embedding_texts:
        return [], np.empty((0, 768), dtype=np.float32), {}
    nutrient_embedding_settings = {
        "model": NUTRIENT_BERT_MODEL, "revision": NUTRIENT_BERT_REVISION,
        "pooling": "last_hidden_mean_excluding_special_and_padding_v1",
        "max_length": NUTRIENT_BERT_MAX_LENGTH,
        "texts": nutrient_embedding_texts,
    }
    nutrient_embedding_digest = sha256(json.dumps(
        nutrient_embedding_settings, sort_keys=True, ensure_ascii=False
    ).encode()).hexdigest()
    nutrient_embedding_cache = nutrient_cache_dir / f"bert_embeddings_{nutrient_embedding_digest[:16]}.npz"

    if nutrient_embedding_cache.is_file():
        with np.load(nutrient_embedding_cache, allow_pickle=False) as cached:
            if cached["settings_sha256"].item() != nutrient_embedding_digest:
                raise ValueError("Embedding cache settings do not match this run.")
            nutrient_text_embeddings = cached["embeddings"]
        if nutrient_text_embeddings.shape != (len(nutrient_embedding_texts), 768):
            raise ValueError("Embedding cache dimensions do not match this run.")
        print(f"Reused {len(nutrient_embedding_texts):,} cached BERT embeddings")
    else:
        nutrient_bert_tokenizer, nutrient_bert_model = load_nutrient_bert()
        nutrient_bert_device = torch.device(
            "cuda" if torch.cuda.is_available() else
            "mps" if torch.backends.mps.is_available() else "cpu"
        )
        nutrient_bert_model = nutrient_bert_model.to(nutrient_bert_device).eval()
        embedding_batches = []
        with torch.inference_mode():
            for start in range(0, len(nutrient_embedding_texts), NUTRIENT_BERT_BATCH_SIZE):
                texts = nutrient_embedding_texts[start:start + NUTRIENT_BERT_BATCH_SIZE]
                encoded = nutrient_bert_tokenizer(
                    texts, padding=True, truncation=False,
                    return_tensors="pt", return_special_tokens_mask=True,
                )
                # Fail instead of silently discarding words from an ingredient.
                if encoded["input_ids"].shape[1] > NUTRIENT_BERT_MAX_LENGTH:
                    raise ValueError("A phrase exceeds NUTRIENT_BERT_MAX_LENGTH; increase it and rerun.")
                specials = encoded.pop("special_tokens_mask").to(nutrient_bert_device)
                encoded = {key: value.to(nutrient_bert_device) for key, value in encoded.items()}
                hidden = nutrient_bert_model(**encoded).last_hidden_state
                content_mask = (encoded["attention_mask"].bool() & ~specials.bool()).unsqueeze(-1)
                pooled = (hidden * content_mask).sum(dim=1) / content_mask.sum(dim=1).clamp(min=1)
                vectors = F.normalize(pooled.float(), p=2, dim=1)
                embedding_batches.append(vectors.cpu().numpy())
                if start == 0 or (start // NUTRIENT_BERT_BATCH_SIZE + 1) % 20 == 0:
                    print(f"Embedded {min(start + len(texts), len(nutrient_embedding_texts)):,}/{len(nutrient_embedding_texts):,} phrases", flush=True)
        nutrient_text_embeddings = np.concatenate(embedding_batches).astype(np.float32)
        np.savez_compressed(
            nutrient_embedding_cache, embeddings=nutrient_text_embeddings,
            settings_sha256=np.array(nutrient_embedding_digest),
        )
        del nutrient_bert_model
        print(f"Saved embeddings to {nutrient_embedding_cache}")
    assert np.isfinite(nutrient_text_embeddings).all()
    assert np.allclose(np.linalg.norm(nutrient_text_embeddings, axis=1), 1, atol=1e-5)
    nutrient_embedding_positions = {text: i for i, text in enumerate(nutrient_embedding_texts)}
    return nutrient_embedding_texts, nutrient_text_embeddings, nutrient_embedding_positions

# Embed EVERY ingredient before extraction or matching, including ingredients
# that will later match exactly. Normalize/singularize and encode duplicate
# keys once; all original ingredient phrases retain an embedding row lookup.
dataset_ingredient_phrases = sorted({
    ingredient for recipe_id in recipe_names
    for ingredient in recipe_ingredients.get(recipe_id, [])
})
ingredient_embedding_df = pd.DataFrame({"ingredient": dataset_ingredient_phrases})
ingredient_embedding_df["embedding_key"] = ingredient_embedding_df["ingredient"].map(normalize_price_text)
if ingredient_embedding_df["embedding_key"].eq("").any():
    raise ValueError("An ingredient normalized to empty text; review it before embedding.")
ingredient_embedding_texts, ingredient_bert_embeddings, ingredient_embedding_positions = (
    get_nutrient_bert_embeddings(ingredient_embedding_df["embedding_key"])
)
ingredient_embedding_df["embedding_index"] = ingredient_embedding_df["embedding_key"].map(ingredient_embedding_positions)
print(f"Embedded all {len(ingredient_embedding_df):,} ingredient phrases as {len(ingredient_embedding_texts):,} distinct normalized vectors.")
```

```python
from scipy.sparse import vstack
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors, sort_graph_by_row_values

INGREDIENT_DBSCAN_EPS = 0.08  # cosine distance: adjacent vectors have similarity >= 0.92
INGREDIENT_DBSCAN_MIN_SAMPLES = 2
INGREDIENT_NEIGHBOR_BATCH_SIZE = 512

# Sparse radius neighborhoods avoid retaining a dense all-pairs distance matrix.
ingredient_neighbor_model = NearestNeighbors(
    radius=INGREDIENT_DBSCAN_EPS, metric="cosine", algorithm="brute"
).fit(ingredient_bert_embeddings)
ingredient_radius_graph = vstack([
    ingredient_neighbor_model.radius_neighbors_graph(
        ingredient_bert_embeddings[start:start + INGREDIENT_NEIGHBOR_BATCH_SIZE], mode="distance"
    )
    for start in range(0, len(ingredient_bert_embeddings), INGREDIENT_NEIGHBOR_BATCH_SIZE)
], format="csr")
ingredient_radius_graph = sort_graph_by_row_values(ingredient_radius_graph, warn_when_not_sorted=False)
ingredient_dbscan = DBSCAN(
    eps=INGREDIENT_DBSCAN_EPS, min_samples=INGREDIENT_DBSCAN_MIN_SAMPLES, metric="precomputed"
).fit(ingredient_radius_graph)
ingredient_cluster_labels = ingredient_dbscan.labels_
ingredient_embedding_df["dbscan_cluster"] = ingredient_cluster_labels[
    ingredient_embedding_df["embedding_index"].to_numpy()
]
ingredient_cluster_lookup = ingredient_embedding_df.set_index("ingredient")
ingredient_cluster_summary = (
    ingredient_embedding_df.loc[ingredient_embedding_df["dbscan_cluster"].ge(0)]
    .groupby("dbscan_cluster")
    .agg(unique_vectors=("embedding_key", "nunique"), ingredient_phrases=("ingredient", "size"),
         examples=("ingredient", lambda values: list(values.head(6))))
    .sort_values(["unique_vectors", "ingredient_phrases"], ascending=False)
)
print(f"DBSCAN: {len(ingredient_cluster_summary):,} clusters; {(ingredient_cluster_labels == -1).sum():,} unclustered normalized ingredients.")
print(f"Original ingredient phrases assigned to clusters: {ingredient_embedding_df['dbscan_cluster'].ge(0).sum():,}/{len(ingredient_embedding_df):,}")

def find_cluster_price_neighbors(query_ingredients, priced_ingredients):
    """Eligible direct-priced neighbors only; never search noise or other clusters."""
    queries = ingredient_cluster_lookup.loc[sorted(set(query_ingredients))]
    anchors = ingredient_cluster_lookup.loc[sorted(set(priced_ingredients))]
    rows = []
    for cluster_id, group in queries.loc[queries["dbscan_cluster"].ge(0)].groupby("dbscan_cluster", sort=True):
        same_cluster = anchors.loc[anchors["dbscan_cluster"].eq(cluster_id)]
        if same_cluster.empty:
            continue
        anchor_names = same_cluster.index.to_numpy()
        anchor_vectors = ingredient_bert_embeddings[same_cluster["embedding_index"].to_numpy()]
        for start in range(0, len(group), 256):
            chunk = group.iloc[start:start + 256]
            query_vectors = ingredient_bert_embeddings[chunk["embedding_index"].to_numpy()]
            scores = np.clip(query_vectors @ anchor_vectors.T, -1.0, 1.0)
            query_rows, anchor_rows = np.where(scores > NUTRIENT_BERT_THRESHOLD)
            rows.extend({
                "ingredient": chunk.index[int(i)], "dbscan_cluster": int(cluster_id),
                "cluster_neighbor": anchor_names[int(j)],
                "cluster_neighbor_cosine": float(scores[i, j]),
            } for i, j in zip(query_rows, anchor_rows)
                if chunk.index[int(i)] != anchor_names[int(j)])
    return pd.DataFrame(rows, columns=[
        "ingredient", "dbscan_cluster", "cluster_neighbor", "cluster_neighbor_cosine",
    ])

ingredient_cluster_summary.head(20)
```

```python
import json
import re
import tempfile
from pathlib import Path
from hashlib import sha256

INGREDIENT_NER_MODEL = "urchade/gliner_large-v2.1"
INGREDIENT_NER_LABELS = ["food", "color", "texture", "preparation", "adjective"]
INGREDIENT_NER_THRESHOLD = 0.3
INGREDIENT_NER_BATCH_SIZE = 64
INGREDIENT_NER_PREFIX = "The recipe calls for "
INGREDIENT_NER_SUFFIX = "."
ingredient_ner_phrases = sorted({
    ingredient for recipe_id in recipe_names
    for ingredient in recipe_ingredients.get(recipe_id, [])
})
ingredient_ner_settings = {
    "model": INGREDIENT_NER_MODEL, "labels": INGREDIENT_NER_LABELS,
    "threshold": INGREDIENT_NER_THRESHOLD, "flat_ner": False,
    "multi_label": True, "prefix": INGREDIENT_NER_PREFIX,
    "suffix": INGREDIENT_NER_SUFFIX, "offset_filter_version": 1,
    "phrases": ingredient_ner_phrases,
}
ingredient_ner_digest = sha256(json.dumps(ingredient_ner_settings, sort_keys=True).encode()).hexdigest()
ingredient_ner_cache_dir = (Path(tempfile.gettempdir()) / "recipe_price_mapping_cache"
    / sha256(str(Path.cwd().resolve()).encode()).hexdigest()[:16])
ingredient_ner_cache_dir.mkdir(parents=True, exist_ok=True)
ingredient_ner_cache = ingredient_ner_cache_dir / f"ingredient_ner_{ingredient_ner_digest[:16]}.json"
if ingredient_ner_cache.is_file():
    ingredient_ner_predictions = json.loads(ingredient_ner_cache.read_text())
    if len(ingredient_ner_predictions) != len(ingredient_ner_phrases):
        raise ValueError("Ingredient NER cache does not align with the phrase list.")
    print(f"Reused GLiNER tags for {len(ingredient_ner_phrases):,} ingredient phrases")
else:
    from gliner import GLiNER
    ingredient_ner_model = GLiNER.from_pretrained(INGREDIENT_NER_MODEL, local_files_only=True)
    ingredient_ner_predictions = []
    for start in range(0, len(ingredient_ner_phrases), INGREDIENT_NER_BATCH_SIZE):
        chunk = ingredient_ner_phrases[start:start + INGREDIENT_NER_BATCH_SIZE]
        context_texts = [INGREDIENT_NER_PREFIX + phrase + INGREDIENT_NER_SUFFIX for phrase in chunk]
        context_predictions = ingredient_ner_model.inference(
            context_texts, INGREDIENT_NER_LABELS, threshold=INGREDIENT_NER_THRESHOLD,
            batch_size=INGREDIENT_NER_BATCH_SIZE, flat_ner=False, multi_label=True,
        )
        # Only keep spans wholly inside the ingredient; convert offsets back to
        # its original text so "recipe", "calls" and the sentence punctuation
        # can never become ingredient entities.
        offset = len(INGREDIENT_NER_PREFIX)
        for phrase, predictions in zip(chunk, context_predictions):
            ingredient_ner_predictions.append([
                {**entity, "start": entity["start"] - offset,
                 "end": entity["end"] - offset,
                 "text": phrase[entity["start"] - offset:entity["end"] - offset]}
                for entity in predictions
                if offset <= entity["start"] < entity["end"] <= offset + len(phrase)
            ])
        if start == 0 or (start // INGREDIENT_NER_BATCH_SIZE + 1) % 10 == 0:
            print(f"Tagged {len(ingredient_ner_predictions):,}/{len(ingredient_ner_phrases):,} ingredients", flush=True)
    ingredient_ner_cache.write_text(json.dumps(ingredient_ner_predictions, ensure_ascii=False))
    del ingredient_ner_model
    print(f"Cached GLiNER tags for {len(ingredient_ner_phrases):,} ingredients")
```

```python
# GLiNER can retain qualifiers inside a food span or miss short preparation
# words. These explicit rules supplement its tags; they are not model output.
# Numeric measures joined to a word, % or a hyphen: "10 inch", "10-inch",
# "1%", "10 bean", "2 oz". These are quantities, not products or modifiers:
# strip them from product spans and never add them to adj/verb.
NUMBER_REGEX = re.compile(
    r"\d+(?:[./]\d+)?(?:%|(?:\s*[-\u2013]\s*|\s+)?[A-Za-z]+)"
)
INGREDIENT_PREPARATION_WORDS = {
    "ground", "minced", "chopped", "diced", "sliced", "peeled", "grated",
    "shredded", "crushed", "mashed", "cooked", "uncooked", "boiled",
    "steamed", "fried", "roasted", "toasted", "baked", "grilled", "smoked",
    "canned", "drained", "rinsed", "sifted", "beaten", "whipped", "melted",
    "softened", "chilled", "frozen", "thawed", "dried", "dehydrated",
    "seeded", "deseeded", "pitted", "shelled", "trimmed", "cubed",
}
INGREDIENT_ADJECTIVE_WORDS = {
    "yellow", "red", "green", "white", "black", "purple", "brown",
    "fresh", "raw", "ripe", "unripe", "boneless", "skinless", "lean",
    "large", "small", "medium", "fine", "coarse", "organic", "unsalted",
    "salted", "sweet", "sour", "hot", "cold", "warm", "extra", "virgin",
}
PROTECTED_INGREDIENT_COMPOUNDS = {
    "sour cream", "sweet potato", "sweet corn", "hot dog", "ice cream",
    "cream cheese", "peanut butter", "baking powder", "baking soda",
    "red wine", "white wine", "brown sugar", "black bean", "green bean",
}

def ingredient_entity_columns(phrase, predictions):
    valid = [entity for entity in predictions
             if 0 <= entity["start"] < entity["end"] <= len(phrase)]
    # Keep the strongest non-overlapping food spans. Nested generic nouns
    # should not split "cream cheese" into separate cream and cheese products.
    foods = []
    for entity in sorted((e for e in valid if e["label"] == "food"),
                         key=lambda e: (-e["score"], -(e["end"] - e["start"]), e["start"])):
        if not any(entity["start"] < old["end"] and old["start"] < entity["end"] for old in foods):
            foods.append(entity)
    foods.sort(key=lambda e: e["start"])
    quantity_spans = [(match.start(), match.end()) for match in NUMBER_REGEX.finditer(phrase)]

    def overlaps_quantity(start, end):
        return any(start < quantity_end and quantity_start < end
                   for quantity_start, quantity_end in quantity_spans)

    modifiers = []
    label_columns = {"color": "adj", "texture": "adj", "adjective": "adj", "preparation": "verb"}
    for entity in valid:
        column = label_columns.get(entity["label"])
        if column is None:
            continue
        # Ignore broad contradictory tags such as the entire "sour cream"
        # labelled texture, or the same "parsley" span labelled food/color.
        if entity["start"] == 0 and entity["end"] == len(phrase):
            continue
        if any(entity["start"] <= food["start"] and entity["end"] >= food["end"] for food in foods):
            continue
        # Quantity spans ("10 inch", "1%") are not adjectives or verbs.
        if overlaps_quantity(entity["start"], entity["end"]):
            continue
        words = list(re.finditer(r"[A-Za-z]+", phrase[entity["start"]:entity["end"]]))
        known_words = INGREDIENT_PREPARATION_WORDS | INGREDIENT_ADJECTIVE_WORDS
        if words and all(word.group().lower() in known_words for word in words):
            for word in words:
                text = word.group().lower()
                modifiers.append((entity["start"] + word.start(), entity["start"] + word.end(),
                                  "verb" if text in INGREDIENT_PREPARATION_WORDS else "adj", text))
        else:
            modifiers.append((entity["start"], entity["end"], column, entity["text"].lower()))
    for word in re.finditer(r"[A-Za-z]+", phrase):
        text = word.group().lower()
        if overlaps_quantity(word.start(), word.end()):
            continue
        if text in INGREDIENT_PREPARATION_WORDS | INGREDIENT_ADJECTIVE_WORDS:
            modifiers.append((word.start(), word.end(),
                              "verb" if text in INGREDIENT_PREPARATION_WORDS else "adj", text))
    # Prefer a preparation verb over an overlapping adjective at the same span.
    distinct = {}
    for start, end, column, text in modifiers:
        key = (start, end)
        if key not in distinct or column == "verb":
            distinct[key] = (start, end, column, text)
    modifiers = sorted(distinct.values())
    result = {"ingredient": phrase, "product": [], "adj": [], "verb": []}
    for _, _, column, text in modifiers:
        if text not in result[column]:
            result[column].append(text)
    for food in foods:
        surface = phrase[food["start"]:food["end"]]
        if normalize_price_text(surface) in PROTECTED_INGREDIENT_COMPOUNDS:
            core = normalize_price_text(surface)
        else:
            chars = list(surface)
            removals = [(start, end) for start, end, _, _ in modifiers] + quantity_spans
            for start, end in removals:
                if food["start"] <= start and end <= food["end"]:
                    chars[start - food["start"]:end - food["start"]] = " " * (end - start)
            core = normalize_price_text("".join(chars))
        if core and core not in result["product"]:
            result["product"].append(core)
    return result

ingredient_entities_df = pd.DataFrame([
    ingredient_entity_columns(phrase, predictions)
    for phrase, predictions in zip(ingredient_ner_phrases, ingredient_ner_predictions)
], columns=["ingredient", "product", "adj", "verb"])
ingredient_entity_lookup = ingredient_entities_df.set_index("ingredient").to_dict("index")
# One entry per ingredient. Compound ingredients retain multiple products.
recipe_ingredient_entities_df = pd.DataFrame([
    {"recipe_id": recipe_id,
     **{column: [ingredient_entity_lookup[item][column] for item in recipe_ingredients.get(recipe_id, [])]
        for column in ["product", "adj", "verb"]}}
    for recipe_id in recipe_names
])
print(f"Extracted products for {ingredient_entities_df['product'].map(bool).sum():,}/{len(ingredient_entities_df):,} unique ingredients")
from IPython.display import display
display(ingredient_entities_df.loc[ingredient_entities_df["ingredient"].isin([
    "yellow onion", "ground beef", "fresh parsley", "cream cheese", "salt and pepper", "canned crushed tomatoes",
    "1% milk", "10 inch flour tortilla", "10 bean soup mix", "10-inch skillet cornbread"
])])
ingredient_entities_df.head(20)
```

```python
def build_fairprice_index(frame):
    frame = frame.copy()
    for field in ["product", "brand", "variant", "origin", "name"]:
        frame[f"{field}_key"] = frame[field].map(normalize_price_text)
    records = frame.to_dict("records")
    suffix_index = defaultdict(list)
    for listing_id, listing in enumerate(records):
        listing["evidence_words"] = set(" ".join(
            listing[f"{field}_key"]
            for field in ["product", "brand", "variant", "origin", "name"]
        ).split())
        words = listing["product_key"].split()
        for start in range(len(words)):
            suffix_index[" ".join(words[start:])].append(listing_id)
    return records, suffix_index

fairprice_records, fairprice_suffix_index = build_fairprice_index(fairprice_df)

# A suffix alone does not establish food identity: ice cream is not cream,
# rice flour is not plain flour, and oyster sauce can end in "reduced salt".
# Only these explicit source modifiers may disappear from the product span.
PRICE_NEUTRAL_MODIFIERS = {
    "fresh", "organic", "premium", "natural", "pure", "frozen", "chilled",
    "loose", "old", "young", "peeled", "chopped", "diced", "sliced", "grated",
}
PRICE_GENERIC_MODIFIERS = {
    "water": {"drinking", "bottle", "distilled", "mineral", "spring", "still"},
    "salt": {"fine", "sea", "rock", "mineral", "himalayan", "pink", "coarse", "iodized", "table"},
    "pepper": {"black", "white", "ground"},
    "sugar": {"white", "cane", "fine", "caster", "granulated", "raw", "brown", "unrefined"},
    "flour": {"plain", "wheat", "white", "all", "purpose"},
    "cream": {"whipping", "thickened", "double", "single", "cooking", "uht", "all", "purpose", "clotted"},
    "milk": {"low", "fat", "full", "cream", "high", "protein", "hi", "cal", "whole", "skim", "skimmed",
             "pasteurized", "pasteurised", "uht", "homogenized", "homogenised", "lactose", "free", "bottle", "100"},
    "butter": {"salted", "unsalted"},
    "olive oil": {"extra", "virgin", "light"},
}
# Also check full names because the cached NER can drop important modifiers.
PRICE_DISTINCT_FOODS = {
    "water": {"sparkling", "soda", "coconut", "lemon", "peach", "flavored", "flavoured"},
    "cream": {"ice", "coconut", "sour"},
    "milk": {"soy", "soya", "almond", "oat", "oats", "coconut", "chocolate", "banana", "strawberry",
             "vanilla", "cookie", "goat", "sheep", "condensed", "evaporated", "cultured", "culture", "powder"},
    "flour": {"rice", "corn", "tapioca", "almond", "coconut", "glutinous", "self", "raising", "bread", "cake", "tempura"},
    "butter": {"peanut", "almond", "cashew", "cocoa", "coconut"},
    "garlic": {"chili", "black", "pickled"},
}
PRICE_PREPARED_MARKERS = {"rte", "cooked", "marinated", "seasoned", "smoked", "flavored", "flavoured"}

def compatible_price_listing(listing, requested, matched_key):
    if not requested <= listing["evidence_words"]:
        return False
    allowed = (requested | PRICE_NEUTRAL_MODIFIERS
               | PRICE_GENERIC_MODIFIERS.get(matched_key, set())
               | set(listing["brand_key"].split()) | set(listing["origin_key"].split()))
    if not set(listing["product_key"].split()) <= allowed:
        return False
    head = matched_key.split()[-1]
    unwanted = (PRICE_DISTINCT_FOODS.get(head, set()) | PRICE_PREPARED_MARKERS) - requested
    if unwanted & listing["evidence_words"]:
        return False
    if "ready to eat" in listing["name_key"] and not {"ready", "eat"} <= requested:
        return False
    return True

def match_ingredient_price(ingredient, records=None, suffix_index=None, price_source="main_crawl"):
    records = fairprice_records if records is None else records
    suffix_index = fairprice_suffix_index if suffix_index is None else suffix_index
    phrase_key = normalize_price_ingredient(ingredient)
    result = {
        "ingredient": ingredient, "ingredient_key": phrase_key,
        "matched_key": None, "match_route": "unmatched",
        "price_route": None, "matched_products": None, "source_names": None,
        "matched_variants": None, "representative_quantity": None,
        "candidate_listings": 0, "priced_listings": 0,
        "pack_price_sgd": None, "price_per_100g_sgd": None,
        "price_per_100ml_sgd": None, "unit_price_status": "unmatched",
        "price_source": None, "matched_search_terms": None,
    }

    def attempt_key(key):
        """Try the longest product suffix of key first, down to the last word."""
        words = key.split()
        if not words:
            return None
        requested = set(words)
        for start in range(len(words)):
            matched_key = " ".join(words[start:])
            candidates = [
                records[i] for i in suffix_index.get(matched_key, [])
                if compatible_price_listing(records[i], requested, matched_key)
            ]
            if not candidates:
                continue

            # Prefer listings explicitly carrying the requested variant/brand.
            # Food words already covered by the product key do not score again.
            qualifiers = requested - set(matched_key.split())
            def variation_rank(listing):
                return tuple(len(qualifiers & set(listing[f"{field}_key"].split()))
                             for field in ["variant", "brand", "origin"])

            best_rank = max(map(variation_rank, candidates))
            selected = ([row for row in candidates if variation_rank(row) == best_rank]
                        if any(best_rank) else candidates)
            # Prefer a named pack; choose the most frequent, then lexical order
            # to make ties deterministic. Prices only average within that pack.
            packs = pd.Series([row["quantity"] for row in selected if row["quantity"]])
            if packs.empty:
                pack = ""
            else:
                counts = packs.value_counts()
                pack = sorted(counts[counts.eq(counts.max())].index)[0]
            selected = [row for row in selected if row["quantity"] == pack]
            average_price = sum(row["price_value"] for row in selected) / len(selected)
            magnitude, dimension = parse_price_quantity(pack)
            exact_product = all(row["product_key"] == key for row in selected)
            route = ("exact_product" if exact_product else
                     "qualified_suffix" if start > 0 else "product_suffix_average")
            updates = dict(
                ingredient_key=key, matched_key=matched_key, match_route=route,
                price_source=price_source,
                matched_search_terms=sorted({str(row["term"]) for row in selected if row.get("term")}),
                price_route="named_variation" if any(best_rank) else "product_average",
                matched_products=sorted({row["product"] for row in selected}),
                source_names=sorted({row["name"] for row in selected}),
                matched_variants=sorted({row["variant"] for row in selected if row["variant"]}),
                representative_quantity=pack,
                candidate_listings=len(candidates), priced_listings=len(selected),
                pack_price_sgd=round(average_price, 4),
                unit_price_status="parsed" if magnitude else "count_only_or_unparsed",
            )
            if magnitude:
                unit_column = "price_per_100g_sgd" if dimension == "g" else "price_per_100ml_sgd"
                updates[unit_column] = round(average_price / magnitude * 100, 6)
            return updates
        return None

    # NER first: a single extracted product is the cleanest search key
    # ("yellow onion" -> "onion", "ground beef" -> "beef"). Phrases with no
    # product or with several products ("salt and pepper"), or whose NER key
    # matches nothing, fall back to the raw normalized phrase suffixes, ending
    # at the last vocabulary word. NER-based routes are prefixed "ner_".
    ner_products = ingredient_entity_lookup.get(ingredient, {}).get("product", [])
    keys_to_try = []
    if len(ner_products) == 1:
        ner_key = normalize_price_text(ner_products[0])
        keys_to_try.append((INGREDIENT_PRICE_ALIASES.get(ner_key, ner_key), "ner_"))
    # A compound/alternative is not one purchasable ingredient.
    if not re.search(r"\b(?:and|or)\b", str(ingredient).lower()):
        keys_to_try.append((phrase_key, ""))
    seen_keys = set()
    for key, route_prefix in keys_to_try:
        if not key or key in seen_keys:
            continue
        seen_keys.add(key)
        updates = attempt_key(key)
        if updates is not None:
            updates["match_route"] = route_prefix + updates["match_route"]
            result.update(updates)
            return result
    return result


# Match each unique name once, instead of once per recipe or rating.
ingredient_frequency = recipe_df["ingredients"].explode().dropna().value_counts()
ingredient_price_df = pd.DataFrame(
    [match_ingredient_price(ingredient) for ingredient in sorted(ingredient_frequency.index)]
)
ingredient_price_df["recipe_occurrences"] = ingredient_price_df["ingredient"].map(ingredient_frequency)
# Keep an unchanged baseline; the fallback fills only missing prices.
primary_ingredient_price_df = ingredient_price_df.copy(deep=True)
ingredient_price_df.head(10)
```

**Second pass: prices from the unmatched-term crawl**

Load `data/output_raw/raw_unmatched_food_prices.csv` after the main crawl.
Extract actual product/brand/variant entities with the reference notebook's
GLiNER model and settings, then use the same normalization, singularization,
suffix lookup and price calculation as the first pass. Search terms themselves
are not treated as product matches: the extracted product head must agree with
the term, as in the reference notebook’s term validation, and non-food results
are excluded. The first run uses the locally cached model;
subsequent runs reuse a temporary `fairprice_unmatched_entities.csv` cache, invalidated
automatically when the source file or extraction settings change.

Only unpriced ingredients are retried, including gaps in partially priced
recipes. Main-crawl prices take precedence. The coverage table below compares
both passes using every key in `recipe_names` as its denominator.

```python
import re

COUNTRIES = {
    "australia": "Australia", "malaysia": "Malaysia", "thailand": "Thailand", "japan": "Japan",
    "china": "China", "korea": "South Korea", "south korea": "South Korea", "taiwan": "Taiwan",
    "vietnam": "Vietnam", "indonesia": "Indonesia", "india": "India", "philippines": "Philippines",
    "singapore": "Singapore", "sg local": "Singapore", "hong kong": "Hong Kong",
    "new zealand": "New Zealand", "usa": "USA", "united states": "USA", "america": "USA",
    "canada": "Canada", "mexico": "Mexico", "peru": "Peru", "chile": "Chile",
    "argentina": "Argentina", "brazil": "Brazil", "ecuador": "Ecuador", "egypt": "Egypt",
    "turkey": "Turkey", "france": "France", "holland": "Netherlands", "netherlands": "Netherlands",
    "italy": "Italy", "spain": "Spain", "germany": "Germany", "denmark": "Denmark",
    "norway": "Norway", "sweden": "Sweden", "iceland": "Iceland", "ireland": "Ireland",
    "scotland": "Scotland", "england": "England", "greece": "Greece", "switzerland": "Switzerland",
    "belgium": "Belgium", "portugal": "Portugal", "poland": "Poland", "russia": "Russia",
    "pakistan": "Pakistan", "myanmar": "Myanmar", "cambodia": "Cambodia", "lebanon": "Lebanon",
    "sri lanka": "Sri Lanka", "kenya": "Kenya", "south africa": "South Africa", "morocco": "Morocco",
}

DEMONYMS = {
    "malaysian": "Malaysia", "australian": "Australia", "norwegian": "Norway", "chilean": "Chile",
    "peruvian": "Peru", "icelandic": "Iceland", "canadian": "Canada", "taiwanese": "Taiwan",
    "vietnamese": "Vietnam", "indonesian": "Indonesia", "indian": "India",
    "korean": "South Korea", "filipino": "Philippines", "japanese": "Japan", "thai": "Thailand",
    "chinese": "China", "greek": "Greece", "french": "France", "italian": "Italy",
    "spanish": "Spain", "german": "Germany", "swiss": "Switzerland", "belgian": "Belgium",
    "danish": "Denmark", "dutch": "Netherlands", "english": "England", "british": "England",
    "scottish": "Scotland", "irish": "Ireland", "american": "USA", "mexican": "Mexico",
    "turkish": "Turkey", "portuguese": "Portugal", "russian": "Russia", "polish": "Poland",
    "swedish": "Sweden", "brazilian": "Brazil", "argentinian": "Argentina",
    "moroccan": "Morocco", "egyptian": "Egypt", "lebanese": "Lebanon",
}

GEO = {**COUNTRIES, **DEMONYMS}

def _pattern(keys):
    # longest first, so "new zealand" wins over a bare "zealand" fragment
    alts = "|".join(sorted(map(re.escape, keys), key=len, reverse=True))
    return re.compile(rf"\b({alts})\b", re.I)

ANY, COUNTRY_ONLY = _pattern(GEO), _pattern(COUNTRIES)
US = re.compile(r"\bU\.?S\.?\b")  # case-sensitive: the pronoun "us" is not a country
MEAT = re.compile(r"\b(ham|breast|sausage|bacon|slice[ds]?|mince[ds]?|roll|frank\w*|deli"
                  r"|drumstick|wing|smoked)\b", re.I)
STYLE = re.compile(r"\b(style|bbq|kimchi|marinated|flavou?r\w*|recipe|seasoning|sauce)\b", re.I)

def find_country(name):
    """(canonical country, the text that matched) for one product name."""
    name = str(name)
    # "Korean Style Fish Cake" names a recipe, so in those rows only the noun forms count.
    pattern = COUNTRY_ONLY if STYLE.search(name) else ANY
    if MEAT.search(name):
        name = re.sub(r"\bturkey\b", " ", name, flags=re.I)  # the bird, not the country
    inside = " ".join(re.findall(r"\(([^)]*)\)", name))
    for text in (inside, name):  # a parenthesised country is the origin marker
        m = pattern.search(text)
        if m:
            return GEO[m.group(1).lower()], m.group(1)
    m = US.search(name)
    return ("USA", m.group(0)) if m else ("", "")

def drop_surface(text, surface):
    """Remove the country the row already matched, so it cannot land in another column."""
    if not surface:
        return text
    out = re.sub(rf"\b{re.escape(surface)}\b", " ", text, flags=re.I)
    out = re.sub(r"\(\s*\)|\[\s*\]", " ", out)
    return re.sub(r"\s{2,}", " ", out).strip(" -")


def build_text(name, category):
    # The category is another signal context for the NER
    prefix = f"Supermarket {str(category).replace('-', ' ')} item: "
    return prefix + str(name), len(prefix)

def best_entity(entities, labels, offset, surface):
    best = {}
    for e in entities:
        if e["start"] < offset:
            continue
        if e["score"] > best.get(e["label"], {"score": -1.0})["score"]:
            best[e["label"]] = e
    # The model folds the country into the brand ("Pasar Australia"), so strip the
    # span this row already matched as its origin.
    return [drop_surface(best[l]["text"], surface) if l in best else "" for l in labels]



from hashlib import sha256

unmatched_price_path = fairprice_path.parent.parent / "output_raw/raw_unmatched_food_prices.csv"
unmatched_raw_df = pd.read_csv(unmatched_price_path)
required_unmatched_columns = {"term", "category", "name", "price_value", "quantity"}
if missing := required_unmatched_columns - set(unmatched_raw_df.columns):
    raise ValueError(f"Unmatched price cache is missing columns: {sorted(missing)}")

# A search term is NOT product identity (e.g. formula results include lotion).
# Tag actual listing names with the SAME GLiNER model, labels, threshold and
# context prefix as 3_process_price.ipynb, then cache these extractions.
FAIRPRICE_NER_MODEL = "urchade/gliner_large-v2.1"
FAIRPRICE_NER_LABELS = ["brand", "variant", "product"]
FAIRPRICE_NER_THRESHOLD = 0.3
# Keep reusable model artifacts outside the two deliverable CSVs.
import tempfile
entity_cache_dir = (Path(tempfile.gettempdir()) / "recipe_price_mapping_cache"
                    / sha256(str(Path.cwd().resolve()).encode()).hexdigest()[:16])
entity_cache_dir.mkdir(parents=True, exist_ok=True)
unmatched_entity_cache = entity_cache_dir / "fairprice_unmatched_entities.csv"
unmatched_entity_metadata = entity_cache_dir / "fairprice_unmatched_entities.meta.json"
entity_cache_signature = {
    "source_sha256": sha256(unmatched_price_path.read_bytes()).hexdigest(),
    "model": FAIRPRICE_NER_MODEL, "labels": FAIRPRICE_NER_LABELS,
    "threshold": FAIRPRICE_NER_THRESHOLD, "extraction_version": 1,
}
cache_valid = (
    unmatched_entity_cache.is_file() and unmatched_entity_metadata.is_file()
    and json.loads(unmatched_entity_metadata.read_text()) == entity_cache_signature
)
if cache_valid:
    unmatched_processed_df = pd.read_csv(unmatched_entity_cache).fillna("")
    print(f"Reused extracted entities for {len(unmatched_processed_df):,} search-crawl listings")
else:
    from gliner import GLiNER
    # The model is already cached locally for the reference notebook.
    # If running on another machine, provision this model/tokenizer first.
    unmatched_ner_model = GLiNER.from_pretrained(FAIRPRICE_NER_MODEL, local_files_only=True)
    unmatched_processed_df = unmatched_raw_df.copy()
    names = unmatched_processed_df["name"].fillna("")
    origins = [find_country(name) for name in names]
    unmatched_processed_df["origin"] = [origin for origin, _ in origins]
    built = [build_text(name, category) for name, category in zip(
        names, unmatched_processed_df["category"].fillna("")
    )]
    entity_rows = []
    batch_size = 16
    for start in range(0, len(built), batch_size):
        chunk = built[start:start + batch_size]
        predictions = unmatched_ner_model.inference(
            [text for text, _ in chunk], FAIRPRICE_NER_LABELS,
            threshold=FAIRPRICE_NER_THRESHOLD, batch_size=batch_size,
            flat_ner=True, multi_label=True,
        )
        entity_rows.extend(
            best_entity(prediction, FAIRPRICE_NER_LABELS, offset, surface)
            for prediction, (_, offset), (_, surface) in zip(
                predictions, chunk, origins[start:start + batch_size]
            )
        )
        print(f"Tagged {min(start + batch_size, len(built))}/{len(built)} listings", flush=True)
    unmatched_processed_df[FAIRPRICE_NER_LABELS] = pd.DataFrame(entity_rows, index=unmatched_processed_df.index)
    unmatched_processed_df.to_csv(unmatched_entity_cache, index=False)
    unmatched_entity_metadata.write_text(json.dumps(entity_cache_signature, indent=2))
    del unmatched_ner_model

# Apply the same cleaning to both crawls; build_fairprice_index applies the
# shared reference normalizer/singularizer to product, name and qualifiers.
for column in ["name", "product", "quantity", "brand", "variant", "origin", "term"]:
    unmatched_processed_df[column] = unmatched_processed_df[column].fillna("").astype(str).str.strip()
unmatched_processed_df["price_value"] = pd.to_numeric(unmatched_processed_df["price_value"], errors="coerce")
unmatched_processed_df = unmatched_processed_df.loc[
    unmatched_processed_df["price_value"].gt(0)
    & unmatched_processed_df["price_value"].lt(float("inf"))
].copy()
unmatched_processed_df["quantity"] = (
    unmatched_processed_df["quantity"].str.lower().str.replace(r"\s+", "", regex=True)
)
unmatched_processed_df["product"] = unmatched_processed_df["product"].where(
    unmatched_processed_df["product"].ne(""), unmatched_processed_df["name"]
)
# Reference term_keys validation: a search result must actually name the
# crawled food as its extracted product head. "Sandwich biscuits" cannot price
# sandwiches, and a popcorn flavour tagged as "Cheddar Cheese" cannot price cheese.
def search_listing_is_on_term(row):
    product_words = normalize_price_text(row["product"]).split()
    term = normalize_price_text(row["term"]).replace(" ", "")
    heads = {"".join(product_words[-width:]) for width in (1, 2)}
    return bool(product_words) and term in heads

on_term = unmatched_processed_df.apply(search_listing_is_on_term, axis=1)
# The search also returned non-food merchandise. Those are not ingredient prices.
nonfood = unmatched_processed_df["name"].str.contains(
    r"lotion|massage|hamburger wrapper|cheeseburger wrapper|with prints", case=False, regex=True
)
print(f"On-term food listings: {(on_term & ~nonfood).sum()}/{len(unmatched_processed_df)}")
unmatched_processed_df = unmatched_processed_df.loc[on_term & ~nonfood].copy()
# Different search terms can return the same listing. Filter before deduplicating
# so an off-term occurrence cannot hide the same listing's on-term occurrence.
unmatched_processed_df = unmatched_processed_df.drop_duplicates(
    subset=["name", "quantity", "price_value"]
).reset_index(drop=True)
unmatched_records, unmatched_suffix_index = build_fairprice_index(unmatched_processed_df)
print(f"Indexed {len(unmatched_records):,} distinct priced search-crawl listings")
```

**Recipe price summaries and coverage**

`pack_basket_sgd` is available only when every ingredient has a pack price.
`matched_pack_subtotal_sgd` is explicitly partial when coverage is below 100%.
The two `matched_100*_basket_sgd` columns sum only ingredients with that unit
basis; their separate counts show what each sum covers. None of these columns
uses the recipe's serving size or ingredient quantities, which are unavailable.

Map recipes before joining ratings so a recipe's popularity cannot multiply its
price. All recipes and ratings are retained, including recipes with no matches.

```python
price_lookup = ingredient_price_df.set_index("ingredient").to_dict("index")

def summarize_recipe_prices(ingredients):
    entries = [price_lookup[ingredient] for ingredient in ingredients]
    priced = [row for row in entries if pd.notna(row["pack_price_sgd"])]
    mass = [row["price_per_100g_sgd"] for row in entries if pd.notna(row["price_per_100g_sgd"])]
    volume = [row["price_per_100ml_sgd"] for row in entries if pd.notna(row["price_per_100ml_sgd"])]
    partial = round(sum(row["pack_price_sgd"] for row in priced), 2) if priced else None
    return {
        "total_price_ingredients": len(entries),
        "priced_ingredients": len(priced),
        "price_coverage": len(priced) / len(entries) if entries else None,
        "unpriced_ingredients": [ingredient for ingredient in ingredients
                                 if pd.isna(price_lookup[ingredient]["pack_price_sgd"])],
        "matched_pack_subtotal_sgd": partial,
        "pack_basket_sgd": partial if entries and len(priced) == len(entries) else None,
        "ingredients_with_100g_price": len(mass),
        "ingredients_with_100ml_price": len(volume),
        "priced_ingredients_without_unit_price": len(priced) - len(mass) - len(volume),
        "matched_100g_basket_sgd": round(sum(mass), 4) if mass else None,
        "matched_100ml_basket_sgd": round(sum(volume), 4) if volume else None,
    }

recipe_price_summary_df = pd.DataFrame([
    {"recipe_id": recipe_id, **summarize_recipe_prices(ingredients)}
    for recipe_id, ingredients in recipe_df[["recipe_id", "ingredients"]].itertuples(index=False)
])
recipe_priced_df = recipe_df.merge(recipe_price_summary_df, on="recipe_id", how="left", validate="one_to_one")
# Keep the detailed unmatched list in recipe_priced_df; avoid repeating it per rating.
final_priced_df = final_df.merge(
    recipe_price_summary_df.drop(columns="unpriced_ingredients"),
    on="recipe_id", how="left", validate="many_to_one",
)
assert len(recipe_priced_df) == len(recipe_df)
assert len(final_priced_df) == len(final_df)
assert recipe_price_summary_df["priced_ingredients"].le(recipe_price_summary_df["total_price_ingredients"]).all()

print(f"Unique ingredients priced: {ingredient_price_df['pack_price_sgd'].notna().sum():,} / {len(ingredient_price_df):,}")
print(f"Ingredient occurrence coverage: {recipe_price_summary_df['priced_ingredients'].sum() / recipe_price_summary_df['total_price_ingredients'].sum():.1%}")
print(f"Fully priced recipes: {recipe_price_summary_df['pack_basket_sgd'].notna().sum():,} / {len(recipe_price_summary_df):,}")
recipe_priced_df[["recipe_id", "name", "priced_ingredients", "total_price_ingredients",
                  "price_coverage", "pack_basket_sgd", "unpriced_ingredients"]].head(10)
```

**Recipe coverage based on `recipe_names`**

Percentages use `len(recipe_names)` as the denominator. “Half matched” means **at least 50%** of a recipe’s ingredients have a matched pack price. These thresholds overlap; fully matched recipes satisfy all three. Recipes with no ingredients or no summary stay in the denominator and do not count as half or fully matched.

```python
# Each recipe_names key counts once, regardless of its number of ratings.
# Missing summary rows remain in the denominator and count as unmatched.
recipe_price_coverage_df = (
    recipe_price_summary_df.set_index("recipe_id")
    .reindex(pd.Index(recipe_names.keys(), name="recipe_id"))
)
matched_counts = recipe_price_coverage_df["priced_ingredients"].fillna(0)
ingredient_counts = recipe_price_coverage_df["total_price_ingredients"].fillna(0)
has_ingredients = ingredient_counts.gt(0)
total_named_recipes = len(recipe_names)

coverage_conditions = {
    "At least one ingredient priced": matched_counts.gt(0),
    "At least half of ingredients priced (>= 50%)": has_ingredients & (matched_counts * 2).ge(ingredient_counts),
    "Fully priced (100%)": has_ingredients & matched_counts.eq(ingredient_counts),
}
recipe_price_coverage_report = pd.DataFrame([
    {
        "Price coverage": label,
        "Recipes": int(condition.sum()),
        "Percentage of recipe_names": round(condition.sum() / total_named_recipes * 100, 2)
            if total_named_recipes else float("nan"),
    }
    for label, condition in coverage_conditions.items()
])
print(f"Denominator: {total_named_recipes:,} recipes in recipe_names")
print("Cumulative thresholds: fully priced recipes also count in the first two rows.")

primary_price_lookup = primary_ingredient_price_df.set_index("ingredient")["pack_price_sgd"].to_dict()
primary_counts = pd.Series({
    recipe_id: sum(pd.notna(primary_price_lookup.get(ingredient)) for ingredient in recipe_ingredients.get(recipe_id, []))
    for recipe_id in recipe_names
}).reindex(recipe_price_coverage_df.index).fillna(0)
primary_conditions = [
    primary_counts.gt(0),
    has_ingredients & (primary_counts * 2).ge(ingredient_counts),
    has_ingredients & primary_counts.eq(ingredient_counts),
]
recipe_price_coverage_report["Main crawl recipes"] = [int(mask.sum()) for mask in primary_conditions]
recipe_price_coverage_report["Main crawl percentage"] = [
    round(mask.sum() / total_named_recipes * 100, 2) if total_named_recipes else float("nan")
    for mask in primary_conditions
]
recipe_price_coverage_report["Additional recipes from fallback"] = (
    recipe_price_coverage_report["Recipes"] - recipe_price_coverage_report["Main crawl recipes"]
)
recipe_price_coverage_report[[
    "Price coverage", "Main crawl recipes", "Main crawl percentage",
    "Recipes", "Percentage of recipe_names", "Additional recipes from fallback",
]]
```

```python
from pathlib import Path
from hashlib import sha256
import ast
import json
import numpy as np
import pandas as pd
from nltk.stem import WordNetLemmatizer

nutrient_price_relative_path = Path(
    "ProjectSem1_1/Workshop-Project-Submission-2026-10-10-AIS/"
    "SystemCode/remy/data-crawler/src/data_crawler/data/output_final/price_mapped_nutrients.csv"
)
nutrient_price_path = next(
    (parent / nutrient_price_relative_path for parent in [Path.cwd(), *Path.cwd().parents]
     if (parent / nutrient_price_relative_path).is_file()), None
)
if nutrient_price_path is None:
    raise FileNotFoundError(f"Cannot find {nutrient_price_relative_path}")
nutrient_description_candidates_df = pd.read_csv(nutrient_price_path)
required = {"fdc_id", "description", "product", "adj", "verb", "total_price", "priced_share"}
if missing := required - set(nutrient_description_candidates_df.columns):
    raise ValueError(f"Nutrient price source is missing columns: {sorted(missing)}")

def parse_entity_vocab(value):
    """Return a clean list from a list value or a CSV stringified list."""
    if isinstance(value, list):
        parsed = value
    elif value is None or (isinstance(value, float) and pd.isna(value)):
        parsed = []
    elif isinstance(value, str):
        try:
            parsed = ast.literal_eval(value)
        except (SyntaxError, ValueError):
            parsed = [value]
    else:
        parsed = [value]
    if not isinstance(parsed, (list, tuple, set)):
        parsed = [parsed]
    return [str(item).strip() for item in parsed if str(item).strip()]

ENTITY_WORDNET_POS = {"product": "n", "adj": "a", "verb": "v"}
entity_lemmatizer = WordNetLemmatizer()
# Fail clearly when the local WordNet corpus is unavailable rather than silently
# calling stemming "lemmatization".
try:
    assert entity_lemmatizer.lemmatize("waffles", "n") == "waffle"
    assert entity_lemmatizer.lemmatize("frozen", "v") == "freeze"
except LookupError as error:
    raise LookupError("NLTK WordNet is required for product/adj/verb lemmatization.") from error

def normalize_entity_vocab(value, field):
    """Parse, normalize, POS-lemmatize, and singularize one entity-list cell."""
    pos = ENTITY_WORDNET_POS[field]
    normalized = []
    for phrase in parse_entity_vocab(value):
        cleaned = normalize_description(phrase)
        lemmas = [
            singularize_word(entity_lemmatizer.lemmatize(token, pos)).lower()
            for token in cleaned.split()
        ]
        key = " ".join(token for token in lemmas if token)
        if key and key not in normalized:
            normalized.append(key)
    return normalized

nutrient_description_candidates_df["description_first_segment"] = (
    nutrient_description_candidates_df["description"].fillna("").str.split(",", n=1).str[0].str.strip()
)
nutrient_description_candidates_df["description_key"] = (
    nutrient_description_candidates_df["description_first_segment"].map(normalize_price_text)
)
nutrient_description_candidates_df = nutrient_description_candidates_df.loc[
    nutrient_description_candidates_df["description_key"].ne("")
].reset_index(drop=True)
nutrient_description_candidates_df["nutrient_source_row"] = nutrient_description_candidates_df.index
for field in ["product", "adj", "verb"]:
    nutrient_description_candidates_df[f"{field}_vocab"] = (
        nutrient_description_candidates_df[field].map(lambda value, field=field: normalize_entity_vocab(value, field))
    )

source_totals = pd.to_numeric(nutrient_description_candidates_df["total_price"], errors="coerce")
source_share = pd.to_numeric(nutrient_description_candidates_df["priced_share"], errors="coerce")
nutrient_description_candidates_df["valid_reference_total_sgd"] = source_totals.where(
    source_share.eq(1) & source_totals.gt(0) & np.isfinite(source_totals)
)
nutrient_description_index_df = (
    nutrient_description_candidates_df.groupby("description_key", sort=True)
    .agg(
        candidate_rows=("fdc_id", "size"),
        fully_priced_rows=("valid_reference_total_sgd", "count"),
        median_reference_total_sgd=("valid_reference_total_sgd", "median"),
    )
    .reset_index()
)
nutrient_target_keys = nutrient_description_index_df["description_key"].tolist()
if not nutrient_target_keys:
    raise ValueError("The nutrient source contains no nonempty first-segment keys.")

# Recipe entity columns are already Python lists; normalize them through the
# exact same lemmatize + singularize function used by the nutrient source.
nutrient_ingredient_frequency = pd.Series(
    [ingredient for recipe_id in recipe_names
     for ingredient in recipe_ingredients.get(recipe_id, [])], dtype="object"
).value_counts()
nutrient_ingredient_matches_df = pd.DataFrame({
    "ingredient": sorted(nutrient_ingredient_frequency.index)
})
nutrient_ingredient_matches_df = nutrient_ingredient_matches_df.merge(
    ingredient_entities_df, on="ingredient", how="left", validate="one_to_one"
)
for field in ["product", "adj", "verb"]:
    nutrient_ingredient_matches_df[f"{field}_vocab"] = (
        nutrient_ingredient_matches_df[field].map(lambda value, field=field: normalize_entity_vocab(value, field))
    )
nutrient_ingredient_matches_df["ingredient_key"] = [
    products[0] if len(products) == 1 else normalize_price_text(ingredient)
    for ingredient, products in zip(
        nutrient_ingredient_matches_df["ingredient"],
        nutrient_ingredient_matches_df["product_vocab"],
    )
]
nutrient_ingredient_matches_df["ingredient_key_source"] = nutrient_ingredient_matches_df["product_vocab"].map(
    lambda products: "lemmatized_product" if len(products) == 1 else "original_phrase_fallback"
)
nutrient_ingredient_matches_df["recipe_occurrences"] = nutrient_ingredient_matches_df["ingredient"].map(
    nutrient_ingredient_frequency
)
target_key_set = set(nutrient_target_keys)
nutrient_ingredient_matches_df["exact_description_key"] = [
    product_key if product_key in target_key_set else original_key if original_key in target_key_set else None
    for product_key, original_key in zip(
        nutrient_ingredient_matches_df["ingredient_key"],
        nutrient_ingredient_matches_df["ingredient"].map(normalize_price_text),
    )
]
nutrient_ingredient_matches_df["exact_match"] = nutrient_ingredient_matches_df["exact_description_key"].notna()
print(f"Source: {len(nutrient_description_candidates_df):,} rows / {len(nutrient_target_keys):,} normalized first segments")
print(f"Exact matches: {nutrient_ingredient_matches_df['exact_match'].sum():,} / {len(nutrient_ingredient_matches_df):,} ingredient names")
print("Distinct lemmatized/singularized vocabulary by field:")
for field in ["product", "adj", "verb"]:
    recipe_terms = {term for values in nutrient_ingredient_matches_df[f"{field}_vocab"] for term in values}
    source_terms = {term for values in nutrient_description_candidates_df[f"{field}_vocab"] for term in values}
    print(f"  {field}: recipe={len(recipe_terms):,}, nutrient={len(source_terms):,}, combined={len(recipe_terms | source_terms):,}")
nutrient_description_index_df.head(10)
```

```python
from scipy.sparse import vstack
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors, sort_graph_by_row_values

ENTITY_FIELDS = ["product", "adj", "verb"]
ENTITY_FIELD_WEIGHTS = {"product": 0.70, "adj": 0.15, "verb": 0.15}
NUTRIENT_ENTITY_DBSCAN_EPS = 0.08
NUTRIENT_ENTITY_DBSCAN_MIN_SAMPLES = 2

# Embed each distinct vocabulary phrase once per semantic field. The recipe and
# nutrient DataFrames share each field's BERT vocabulary/coordinate space.
entity_vocab_texts = {}
entity_vocab_embeddings = {}
entity_vocab_positions = {}
for field in ENTITY_FIELDS:
    combined_vocab = {
        term
        for frame in [nutrient_ingredient_matches_df, nutrient_description_candidates_df]
        for values in frame[f"{field}_vocab"]
        for term in values
    }
    texts, raw_vectors, positions = get_nutrient_bert_embeddings(combined_vocab)
    # Vanilla BERT has a strong shared direction: without removing it, unrelated
    # foods can have cosine similarity above 0.8. Center over the vocabulary from
    # both DataFrames, then renormalize so the requested threshold is meaningful.
    centered_vectors = raw_vectors - raw_vectors.mean(axis=0, keepdims=True)
    centered_norms = np.linalg.norm(centered_vectors, axis=1, keepdims=True)
    if not np.isfinite(centered_norms).all() or (centered_norms == 0).any():
        raise ValueError(f"Invalid centered {field} BERT embeddings")
    vectors = (centered_vectors / centered_norms).astype(np.float32)
    entity_vocab_texts[field] = texts
    entity_vocab_embeddings[field] = vectors
    entity_vocab_positions[field] = positions
    print(f"{field}: embedded and mean-centered {len(texts):,} shared vocabulary phrases")

def mean_entity_field_embedding(vocab, field):
    """L2-normalized mean of a row's vocabulary embeddings; None if empty."""
    if not vocab:
        return None
    vectors = entity_vocab_embeddings[field][
        [entity_vocab_positions[field][term] for term in vocab]
    ]
    mean = vectors.mean(axis=0)
    norm = np.linalg.norm(mean)
    if not np.isfinite(norm) or norm == 0:
        raise ValueError(f"Invalid {field} embedding for vocabulary {vocab!r}")
    return (mean / norm).astype(np.float32)

recipe_entity_embeddings_df = nutrient_ingredient_matches_df.copy(deep=True)
nutrient_entity_embeddings_df = nutrient_description_candidates_df.copy(deep=True)
for field in ENTITY_FIELDS:
    recipe_entity_embeddings_df[f"{field}_embedding"] = [
        mean_entity_field_embedding(vocab, field)
        for vocab in recipe_entity_embeddings_df[f"{field}_vocab"]
    ]
    nutrient_entity_embeddings_df[f"{field}_embedding"] = [
        mean_entity_field_embedding(vocab, field)
        for vocab in nutrient_entity_embeddings_df[f"{field}_vocab"]
    ]

# Cluster unique product-list representations from BOTH DataFrames. Deduping
# prevents repeated nutrient rows or popular recipe phrases from changing
# DBSCAN density merely because they occur more often.
product_signatures = sorted({
    tuple(vocab)
    for frame in [recipe_entity_embeddings_df, nutrient_entity_embeddings_df]
    for vocab in frame["product_vocab"]
    if vocab
})
product_signature_vectors = np.stack([
    mean_entity_field_embedding(list(signature), "product")
    for signature in product_signatures
])
product_neighbor_model = NearestNeighbors(
    radius=NUTRIENT_ENTITY_DBSCAN_EPS, metric="cosine", algorithm="brute"
).fit(product_signature_vectors)
product_radius_graph = vstack([
    product_neighbor_model.radius_neighbors_graph(
        product_signature_vectors[start:start + INGREDIENT_NEIGHBOR_BATCH_SIZE], mode="distance"
    )
    for start in range(0, len(product_signature_vectors), INGREDIENT_NEIGHBOR_BATCH_SIZE)
], format="csr")
product_radius_graph = sort_graph_by_row_values(product_radius_graph, warn_when_not_sorted=False)
product_dbscan = DBSCAN(
    eps=NUTRIENT_ENTITY_DBSCAN_EPS,
    min_samples=NUTRIENT_ENTITY_DBSCAN_MIN_SAMPLES,
    metric="precomputed",
).fit(product_radius_graph)
product_signature_cluster = dict(zip(product_signatures, product_dbscan.labels_))
for frame in [recipe_entity_embeddings_df, nutrient_entity_embeddings_df]:
    frame["dbscan_cluster"] = [
        int(product_signature_cluster.get(tuple(vocab), -1))
        for vocab in frame["product_vocab"]
    ]

nutrient_ingredient_matches_df["dbscan_cluster"] = recipe_entity_embeddings_df["dbscan_cluster"]
product_cluster_summary = (
    pd.DataFrame({"product_signature": product_signatures, "dbscan_cluster": product_dbscan.labels_})
    .loc[lambda frame: frame["dbscan_cluster"].ge(0)]
    .groupby("dbscan_cluster")
    .agg(product_representations=("product_signature", "size"),
         examples=("product_signature", lambda values: [" + ".join(value) for value in values.head(6)]))
    .sort_values("product_representations", ascending=False)
)
print(
    f"Entity-product DBSCAN: {len(product_cluster_summary):,} clusters; "
    f"{(product_dbscan.labels_ == -1).sum():,} noise product representations."
)
display(product_cluster_summary.head(20))
```

```python
# Direct matches remain unchanged. Fuzzy candidates are positively priced
# nutrient rows in the same combined product-embedding DBSCAN cluster.
reference_price_lookup = nutrient_description_index_df.set_index("description_key")["median_reference_total_sgd"]
source_is_priced = (
    nutrient_entity_embeddings_df["valid_reference_total_sgd"].gt(0)
    & nutrient_entity_embeddings_df["dbscan_cluster"].ge(0)
    & nutrient_entity_embeddings_df["product_embedding"].notna()
)
priced_nutrient_entities_df = nutrient_entity_embeddings_df.loc[source_is_priced].copy()

unmatched_queries = recipe_entity_embeddings_df.loc[
    ~recipe_entity_embeddings_df["exact_match"]
    & recipe_entity_embeddings_df["dbscan_cluster"].ge(0)
    & recipe_entity_embeddings_df["product_embedding"].notna()
]
sources_by_cluster = {
    int(cluster): group.reset_index(drop=True)
    for cluster, group in priced_nutrient_entities_df.groupby("dbscan_cluster", sort=False)
}
queries_by_cluster = {
    int(cluster): group.reset_index(drop=True)
    for cluster, group in unmatched_queries.groupby("dbscan_cluster", sort=False)
}

# Score whole query/source blocks with matrix multiplication. Product is always
# required; adj/verb contribute only for pairs where both rows provide them.
semantic_rows = []
for cluster_id, query_group in queries_by_cluster.items():
    cluster_sources = sources_by_cluster.get(cluster_id)
    if cluster_sources is None or cluster_sources.empty:
        continue
    source_field_vectors = {}
    source_field_present = {}
    for field in ENTITY_FIELDS:
        raw_vectors = cluster_sources[f"{field}_embedding"].tolist()
        present = np.array([vector is not None for vector in raw_vectors], dtype=bool)
        matrix = np.stack([
            vector if vector is not None else np.zeros(768, dtype=np.float32)
            for vector in raw_vectors
        ])
        source_field_vectors[field] = matrix
        source_field_present[field] = present

    for start in range(0, len(query_group), 256):
        query_chunk = query_group.iloc[start:start + 256]
        query_product = np.stack(query_chunk["product_embedding"].to_numpy())
        product_scores = np.clip(
            query_product @ source_field_vectors["product"].T, -1.0, 1.0
        )
        eligible = product_scores > NUTRIENT_BERT_THRESHOLD
        weighted_sum = ENTITY_FIELD_WEIGHTS["product"] * product_scores
        available_weight = np.full(product_scores.shape, ENTITY_FIELD_WEIGHTS["product"], dtype=np.float32)
        optional_scores = {}

        for field in ["adj", "verb"]:
            raw_query_vectors = query_chunk[f"{field}_embedding"].tolist()
            query_present = np.array([vector is not None for vector in raw_query_vectors], dtype=bool)
            query_matrix = np.stack([
                vector if vector is not None else np.zeros(768, dtype=np.float32)
                for vector in raw_query_vectors
            ])
            scores = np.clip(query_matrix @ source_field_vectors[field].T, -1.0, 1.0)
            pair_has_field = query_present[:, None] & source_field_present[field][None, :]
            optional_scores[field] = scores
            weighted_sum += ENTITY_FIELD_WEIGHTS[field] * scores * pair_has_field
            available_weight += ENTITY_FIELD_WEIGHTS[field] * pair_has_field

        combined_scores = weighted_sum / available_weight
        combined_scores[~eligible] = -np.inf
        candidate_counts = eligible.sum(axis=1)
        best_positions = combined_scores.argmax(axis=1)

        for local_position, query in enumerate(query_chunk.itertuples(index=False)):
            candidate_count = int(candidate_counts[local_position])
            if candidate_count == 0:
                continue
            best_position = int(best_positions[local_position])
            source = cluster_sources.iloc[best_position]
            finite_scores = combined_scores[local_position, eligible[local_position]]
            if candidate_count > 1:
                second = float(np.partition(finite_scores, -2)[-2])
            else:
                second = np.nan
            best_score = float(combined_scores[local_position, best_position])
            adj_score = (
                float(optional_scores["adj"][local_position, best_position])
                if getattr(query, "adj_embedding") is not None
                and source["adj_embedding"] is not None else np.nan
            )
            verb_score = (
                float(optional_scores["verb"][local_position, best_position])
                if getattr(query, "verb_embedding") is not None
                and source["verb_embedding"] is not None else np.nan
            )
            semantic_rows.append({
                "ingredient": query.ingredient,
                "nearest_description_key": source["description_key"],
                "dbscan_cluster": int(cluster_id),
                "cluster_neighbor": " + ".join(source["product_vocab"]),
                "cluster_neighbor_cosine": float(product_scores[local_position, best_position]),
                "product_cosine_similarity": float(product_scores[local_position, best_position]),
                "adj_cosine_similarity": adj_score,
                "verb_cosine_similarity": verb_score,
                "cosine_similarity": best_score,
                "matched_product_vocab": source["product_vocab"],
                "matched_adj_vocab": source["adj_vocab"],
                "matched_verb_vocab": source["verb_vocab"],
                "matched_fdc_id": source["fdc_id"],
                "candidate_count": candidate_count,
                "second_best_similarity": second,
                "similarity_margin": float(best_score - second) if pd.notna(second) else np.nan,
            })

nutrient_semantic_candidates_df = pd.DataFrame(semantic_rows)
semantic_columns = [
    "ingredient", "nearest_description_key", "dbscan_cluster", "cluster_neighbor",
    "cluster_neighbor_cosine", "product_cosine_similarity", "adj_cosine_similarity",
    "verb_cosine_similarity", "cosine_similarity", "matched_product_vocab",
    "matched_adj_vocab", "matched_verb_vocab", "matched_fdc_id", "candidate_count",
    "second_best_similarity", "similarity_margin",
]
if nutrient_semantic_candidates_df.empty:
    nutrient_semantic_candidates_df = pd.DataFrame(columns=semantic_columns)

nutrient_description_matches_df = nutrient_ingredient_matches_df.merge(
    nutrient_semantic_candidates_df.drop(columns="dbscan_cluster"),
    on="ingredient", how="left", validate="one_to_one",
)
exact = nutrient_description_matches_df["exact_match"]
semantic = (
    ~exact
    & nutrient_description_matches_df["product_cosine_similarity"].gt(NUTRIENT_BERT_THRESHOLD)
    & nutrient_description_matches_df["cosine_similarity"].gt(NUTRIENT_BERT_THRESHOLD)
)
nutrient_description_matches_df["match_method"] = np.select(
    [exact, semantic], ["exact_first_segment", "bert_entity_dbscan"], default="unmatched"
)
nutrient_description_matches_df["matched_description_key"] = (
    nutrient_description_matches_df["nearest_description_key"].where(semantic)
)
nutrient_description_matches_df.loc[exact, "matched_description_key"] = (
    nutrient_description_matches_df.loc[exact, "exact_description_key"]
)
nutrient_description_matches_df.loc[exact, "nearest_description_key"] = (
    nutrient_description_matches_df.loc[exact, "exact_description_key"]
)
for score_column in ["product_cosine_similarity", "cosine_similarity"]:
    nutrient_description_matches_df.loc[exact, score_column] = 1.0
nutrient_description_matches_df["candidate_count"] = (
    nutrient_description_matches_df["candidate_count"].fillna(0).astype(int)
)
nutrient_description_matches_df["fallback_status"] = np.select(
    [exact, semantic, nutrient_description_matches_df["dbscan_cluster"].eq(-1),
     nutrient_description_matches_df["candidate_count"].eq(0)],
    ["exact_match", "accepted_entity_match", "dbscan_noise", "no_close_priced_product"],
    default="below_weighted_similarity_threshold",
)
nutrient_description_matches_df = nutrient_description_matches_df.merge(
    nutrient_description_index_df.rename(columns={"description_key": "matched_description_key"}),
    on="matched_description_key", how="left", validate="many_to_one",
)
nutrient_description_matches_df["semantic_review_required"] = (
    nutrient_description_matches_df["match_method"].eq("bert_entity_dbscan")
)
print(nutrient_description_matches_df["match_method"].value_counts().to_string())
print(nutrient_description_matches_df["fallback_status"].value_counts().to_string())
nutrient_description_matches_df.head(15)
```

```python
# Refresh derived fields cleanly even when final_df was loaded from a prior
# CSV export that already includes entity / price columns.
RECIPE_DERIVED_COLUMNS = {
    "total_prices", "ingredient_prices", "price_breakdown",
    "ingredient_semantic_matches", "ingredient_exact_matches", "product", "adj", "verb",
}

def base_recipe_rating_columns(frame):
    return frame.loc[:, [
        column for column in frame.columns
        if re.sub(r"(?:_[xy])+$", "", column) not in RECIPE_DERIVED_COLUMNS
    ]].copy()

base_recipe_ratings_df = base_recipe_rating_columns(final_df)
if "date" not in base_recipe_ratings_df:
    raise ValueError("Rerun the preprocessing cells to carry RAW_interactions.csv dates into final_df.")
# Both exports retain this event-level date through their recipe feature joins.

# One price per ingredient, in the same order as the original ingredients
# list. Unmatched ingredients or matches without a usable price get None.
# The CSV stores this list as JSON, so None is written as null.
def lookup_ingredient_prices(ingredients):
    """Return match rows, prices and a match-aware price breakdown.

    Exact match: ingredient -> price. Fuzzy match:
    ingredient -> {cluster_neighbor: price}. Unmatched: ingredient -> None.
    """
    ingredients = set(ingredients)
    matches = [nutrient_match_lookup[ingredient] for ingredient in ingredients]
    prices = [
        float(row["median_reference_total_sgd"])
        if pd.notna(row["median_reference_total_sgd"]) and row["median_reference_total_sgd"] > 0
        else None
        for row in matches
    ]
    breakdown = {}
    for ingredient, row, price in zip(ingredients, matches, prices):
        if price is None:
            breakdown.setdefault(ingredient, None)
        elif row["match_method"] == "bert_entity_dbscan":
            fuzzy_ingredient = row["cluster_neighbor"]
            if not isinstance(fuzzy_ingredient, str) or not fuzzy_ingredient:
                raise ValueError(f"Fuzzy match for {ingredient!r} has no cluster neighbor.")
            breakdown[ingredient] = {fuzzy_ingredient: price}
        else:
            breakdown[ingredient] = price
    return matches, prices, breakdown

recipe_price_export_rows = []
for recipe_id in recipe_names:
    ingredients = recipe_ingredients.get(recipe_id, [])
    matches, ingredient_prices, price_breakdown = lookup_ingredient_prices(ingredients)
    priced_matches = [row for row, price in zip(matches, ingredient_prices) if price is not None]
    if len(priced_matches) * 2 <= len(ingredients):
        continue  # strictly >50%; excludes empty and exactly-half recipes
    exact_count = sum(row["match_method"] == "exact_first_segment" for row in priced_matches)
    semantic_count = sum(row["match_method"] == "bert_entity_dbscan" for row in priced_matches)
    assert exact_count + semantic_count == len(priced_matches)
    assert len(ingredient_prices) == len(ingredients)
    recipe_price_export_rows.append({
        "recipe_id": recipe_id,
        "total_prices": round(sum(price for price in ingredient_prices if price is not None), 2),
        # Serialize once per recipe, before its user-rating rows repeat it.
        "ingredient_prices": json.dumps(ingredient_prices, allow_nan=False),

        "price_breakdown": json.dumps(price_breakdown, allow_nan=False),
        "ingredient_semantic_matches": semantic_count,
        "ingredient_exact_matches": exact_count,
        **{column: json.dumps([ingredient_entity_lookup[item][column] for item in ingredients])
           for column in ["product", "adj", "verb"]},
    })
recipe_price_export_df = pd.DataFrame(recipe_price_export_rows, columns=[
    "recipe_id", "total_prices", "ingredient_prices", "price_breakdown",
    "ingredient_semantic_matches", "ingredient_exact_matches", "product", "adj", "verb",
])
# final_df is the original recipe/rating data, before any pricing columns.
final_with_prices_df = base_recipe_ratings_df.merge(
    recipe_price_export_df, on="recipe_id", how="inner", validate="many_to_one", sort=False
)
expected_rows = base_recipe_ratings_df["recipe_id"].isin(recipe_price_export_df["recipe_id"]).sum()
assert len(final_with_prices_df) == expected_rows
assert final_with_prices_df["total_prices"].gt(0).all()

price_export_path = Path("input_stage2/recipe_user_ratings_prices.csv")
price_export_path.parent.mkdir(parents=True, exist_ok=True)
temporary_price_export = price_export_path.with_suffix(".csv.tmp")
try:
    final_with_prices_df.to_csv(temporary_price_export, index=False)
    temporary_price_export.replace(price_export_path)
finally:
    if temporary_price_export.exists():
        temporary_price_export.unlink()
print(f"Saved {len(final_with_prices_df):,} rating rows for {final_with_prices_df['recipe_id'].nunique():,} recipes with >50% priced ingredients.")
print("Added columns: total_prices, ingredient_prices, price_breakdown, ingredient_semantic_matches, ingredient_exact_matches, product, adj, verb.")
print("price_breakdown stores entity-BERT fuzzy matches as ingredient: {matched product: price}; JSON null becomes None after json.loads().")

# Show every exported rating row that used at least one fuzzy price match.
# fuzzy_price_recipe_rows_df is also available when a one-row-per-recipe
# view is useful; the displayed table below retains all rating rows.
fuzzy_price_rows_df = final_with_prices_df.loc[
    final_with_prices_df["ingredient_semantic_matches"].gt(0)
].copy()
fuzzy_price_rows_df["fuzzy_price_matches"] = fuzzy_price_rows_df["price_breakdown"].map(
    lambda payload: {
        ingredient: match
        for ingredient, match in json.loads(payload).items()
        if isinstance(match, dict)
    }
)
fuzzy_price_recipe_rows_df = (
    fuzzy_price_rows_df.drop_duplicates("recipe_id").reset_index(drop=True)
)
print(
    f"Rows with at least one fuzzy price match: {len(fuzzy_price_rows_df):,} rating rows / "
    f"{len(fuzzy_price_recipe_rows_df):,} recipes"
)
display(fuzzy_price_rows_df[[
    "recipe_id", "name", "ingredients", "total_prices",
    "ingredient_semantic_matches", "price_breakdown", "fuzzy_price_matches",
]])

# The base recipe/rating CSV gets the same three entity columns plus
# price_breakdown: one JSON value per recipe, aligned with ingredients.
# Unlike the priced CSV, every recipe is kept here, including those below
# the >50% priced threshold.
recipe_entity_export_df = recipe_ingredient_entities_df.copy(deep=True)
for column in ["product", "adj", "verb"]:
    recipe_entity_export_df[column] = recipe_entity_export_df[column].map(json.dumps)
# Same accumulated price_breakdown as the priced CSV, but for every recipe,
# including those below the >50% priced-ingredients threshold.
recipe_entity_export_df["price_breakdown"] = [
    json.dumps(lookup_ingredient_prices(recipe_ingredients.get(recipe_id, []))[2], allow_nan=False)
    for recipe_id in recipe_entity_export_df["recipe_id"]
]
final_with_entities_df = base_recipe_ratings_df.merge(
    recipe_entity_export_df, on="recipe_id", how="left", validate="many_to_one", sort=False
)
assert len(final_with_entities_df) == len(base_recipe_ratings_df)

ratings_export_path = Path("input_stage2/recipe_user_ratings.csv")
temporary_ratings_export = ratings_export_path.with_suffix(".csv.tmp")
try:
    final_with_entities_df.to_csv(temporary_ratings_export, index=False)
    temporary_ratings_export.replace(ratings_export_path)
finally:
    if temporary_ratings_export.exists():
        temporary_ratings_export.unlink()
print(f"Saved {len(final_with_entities_df):,} rating rows for {final_with_entities_df['recipe_id'].nunique():,} recipes to {ratings_export_path}.")
print("Added columns to recipe_user_ratings.csv: product, adj, verb, price_breakdown (JSON, aligned with ingredients).")
```

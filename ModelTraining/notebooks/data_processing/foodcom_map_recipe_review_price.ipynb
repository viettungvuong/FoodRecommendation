```python
import pandas as pd
```

```python
recipes_ingredient_df = pd.read_csv("../../input_stage2/foodcom_recipes_ingredients.csv")
recipes_ingredient_df
```

```python
recipes_user_ratings_df = pd.read_csv("../../input_stage2/recipe_user_ratings.csv")
recipes_user_ratings_df
```

```python
# 7,841 ids repeat in the ingredients file; keep one row per recipe so the join does not duplicate ratings
recipes_ingredient_unique_df = recipes_ingredient_df.drop_duplicates(subset="id", keep="first")

# one row per rating, with its recipe's ingredients; ratings whose recipe is not in the ingredients file are dropped
recipe_ratings_ingredients_df = recipes_user_ratings_df.merge(
    recipes_ingredient_unique_df,
    left_on="recipe_id",
    right_on="id",
    how="inner",
    suffixes=("_rating", "_ingredient"),
).drop(columns="id")
# of the columns both files have (name, ingredients), keep the ingredients file's, under their plain names
# => remove those with _rating suffix
recipe_ratings_ingredients_df = recipe_ratings_ingredients_df.drop(
    columns=[column for column in recipe_ratings_ingredients_df.columns if column.endswith("_rating")]
).rename(columns=lambda column: column.removesuffix("_ingredient"))
percent = len(recipe_ratings_ingredients_df) / len(recipes_user_ratings_df)
print(f"{len(recipe_ratings_ingredients_df):,} of {len(recipes_user_ratings_df):,} ratings matched a recipe with {percent * 100}%")
recipe_ratings_ingredients_df
```

```python
import json

def ingredient_price(price):
    """One ingredient's SGD price: a number, null when unpriced, or {matched term: price} when it was priced through a
    broader term."""
    if isinstance(price, dict):
        found = [value for value in price.values() if value is not None]
        return sum(found) if found else None
    return price

def price_summary(breakdown):
    """A recipe's price_breakdown (ingredient → price) → (total of the priced ingredients, coverage)."""
    prices = [ingredient_price(price) for price in json.loads(breakdown).values()] if isinstance(breakdown, str) else []
    priced = [price for price in prices if price is not None]
    coverage = "none" if not priced else "full" if len(priced) == len(prices) else "partial"
    return round(sum(priced), 2), coverage

# many ratings share a recipe, so each distinct breakdown is parsed once
summaries = {breakdown: price_summary(breakdown) for breakdown in recipe_ratings_ingredients_df["price_breakdown"].unique()}
recipe_ratings_ingredients_df["total_price"] = recipe_ratings_ingredients_df["price_breakdown"].map(lambda b: summaries[b][0])
# full: every ingredient priced; partial: some, so total_price is a lower bound; none: no ingredient priced
recipe_ratings_ingredients_df["price_coverage"] = recipe_ratings_ingredients_df["price_breakdown"].map(lambda b: summaries[b][1])
coverage = recipe_ratings_ingredients_df.drop_duplicates("recipe_id")["price_coverage"]
pd.DataFrame({"count": coverage.value_counts(), "percent": coverage.value_counts(normalize=True).mul(100).round(2)})
```

```python
recipe_ratings_ingredients_df.to_csv("../../input_stage2/foodcom_recipe_ratings_ingredients.csv")
```
